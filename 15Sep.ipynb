```python
# AES 256
!echo "Encryption" > text.txt
!openssl enc -aes-256-cbc -salt -pbkdf2 -in text.txt -out encrypt.bin -pass pass:password
!od -A x -t x1z encrypt.bin | head
!openssl enc -aes-256-cbc -d -pbkdf2 -in encrypt.bin -out decrypt.txt -pass pass:password
!cat decrypt.txt
```

```python
#RSA 2048
!openssl genrsa -out rsa_private.pem 2048
!openssl rsa -in rsa_private.pem -pubout -out rsa_public.pem
!cat rsa_private.pem
!cat rsa_public.pem
!echo "RSA encryption" > rsa_plain.txt
!openssl pkeyutl -encrypt -pubin -inkey rsa_public.pem -in rsa_plain.txt -out rsa_encrypted.bin -pkeyopt rsa_padding_mode:oaep
!openssl pkeyutl -decrypt -inkey rsa_private.pem -in rsa_encrypted.bin -out rsa_decrypted.txt -pkeyopt rsa_padding_mode:oaep
print("Nội dung ban đầu:")
!cat rsa_plain.txt
print("Nội dung đã được giải mã:")
!cat rsa_decrypted.txt
```

```python
# Ký và xác thực với RSA
!echo "Chu ky" > sign.txt
!openssl dgst -sha256 -sign rsa_private.pem -out rsa_signature.bin sign.txt
!openssl dgst -sha256 -verify rsa_public.pem -signature rsa_signature.bin sign.txt
```
